# Data ***Cleaning***

Handle Missing Values

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Amit", "Rahul", "Priya"],
    "Marks": [85, np.nan, 90]
}

df = pd.DataFrame(data)

print("Before Cleaning:")
print(df)

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print("After Cleaning:")
print(df)
```

Remove Duplicate Values

```python
import pandas as pd

data = {
    "Name": ["Amit", "Rahul", "Amit", "Priya"],
    "Marks": [85, 90, 85, 88]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

df = df.drop_duplicates()

print("After Removing Duplicates:")
print(df)
```

Clean Text Data

```python
import pandas as pd

data = {
    "Name": [" amit ", "RAHUL", " Priya "]
}

df = pd.DataFrame(data)

df["Name"] = df["Name"].str.strip().str.title()

print(df)
```

Find Outliers

```python
import pandas as pd

data = {
    "Marks": [45, 48, 50, 52, 55, 57, 60, 99]
}

df = pd.DataFrame(data)

Q1 = df["Marks"].quantile(0.25)
Q3 = df["Marks"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Marks"] < lower) |
    (df["Marks"] > upper)
]

print("Outliers:")
print(outliers)
```

Extract API Data

```python
import requests

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

users = response.json()

for user in users:

    print("Name:", user["name"])
    print("Email:", user["email"])
    print("City:", user["address"]["city"])

    print("-------------------")
```

 API Data into Pandas DataFrame

```python
import requests
import pandas as pd

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

users = response.json()

data = []

for user in users:

    data.append({
        "Name": user["name"],
        "Email": user["email"],
        "City": user["address"]["city"]
    })

df = pd.DataFrame(data)

print(df)

df.to_csv("api_data.csv", index=False)
```

# ***Final Project***

Python Data Analysis Application
Project: Student Performance Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create Dataset

data = {
    "Name": ["Amit", "Rahul", "Priya", "Sneha", "Raj"],
    "StudyHours": [2, 3, 5, 4, 1],
    "Marks": [55, 65, 88, 76, 40]
}

df = pd.DataFrame(data)

# Display Dataset

print("Student Dataset:")
print(df)

# Data Analysis

print("\nAverage Marks:")
print(df["Marks"].mean())

print("\nHighest Marks:")
print(df["Marks"].max())

print("\nLowest Marks:")
print(df["Marks"].min())

# Add Result Column

df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)

print("\nStudent Results:")
print(df)

# Data Visualization

plt.bar(df["Name"], df["Marks"])

plt.title("Student Performance Analysis")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()

# Generate CSV Report

df.to_csv("student_report.csv", index=False)

print("Report Generated Successfully!")
```
